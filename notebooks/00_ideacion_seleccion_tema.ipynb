### Identificación de los 2 mejores temas

Vamos a calcular la puntuación promedio de cada tema y subtema en todos los aspectos de evaluación para identificar los dos temas con mejor calificación general.

```python
# Calcular el promedio de puntuación por subtema a través de todos los aspectos
df_ideas['Puntuacion_Total_Subtema'] = df_ideas[aspectos].mean(axis=1)

# Calcular el promedio de puntuación por Tema (sumando las puntuaciones de sus subtemas)
df_temas_eval = df_ideas.groupby('Tema')['Puntuacion_Total_Subtema'].mean().reset_index()

df_temas_eval = df_temas_eval.sort_values(by='Puntuacion_Total_Subtema', ascending=False)

print("Evaluación promedio por Tema:")
display(df_temas_eval)

print("\nLos 2 mejores temas son:")
display(df_temas_eval.head(2))
```
